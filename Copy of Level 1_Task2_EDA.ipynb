## Conclusion

Exploratory Data Analysis was performed on the Customer Churn dataset using statistical analysis and data visualizations. The analysis helped identify important patterns, relationships, and trends in the data and provided useful insights for further analysis and machine learning.


```python
import pandas as pd

df = pd.read_csv('/content/churn_cleaned_task1.csv')

df.head()
```

```python
numerical_df = df.select_dtypes(include='number')

print("Mean:")
print(numerical_df.mean())

print("\nMedian:")
print(numerical_df.median())

print("\nMode:")
print(numerical_df.mode().iloc[0])

print("\nStandard Deviation:")
print(numerical_df.std())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

numerical_df.hist(figsize=(16, 12), bins=20)

plt.suptitle("Distribution of Numerical Features", fontsize=16)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(16, 10))

sns.boxplot(data=numerical_df)

plt.title("Boxplots of Numerical Features")
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 6))

sns.scatterplot(
    data=df,
    x='Total day minutes',
    y='Total day charge'
)

plt.title('Total Day Minutes vs Total Day Charge')
plt.xlabel('Total Day Minutes')
plt.ylabel('Total Day Charge')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 6))

sns.scatterplot(
    data=df,
    x='Total eve minutes',
    y='Total eve charge'
)

plt.title('Total Evening Minutes vs Total Evening Charge')
plt.xlabel('Total Evening Minutes')
plt.ylabel('Total Evening Charge')
plt.tight_layout()
plt.show()
```

```python
correlation_matrix = numerical_df.corr()

plt.figure(figsize=(14, 10))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    linewidths=0.5
)

plt.title("Correlation Between Numerical Features")
plt.tight_layout()
plt.show()
```
